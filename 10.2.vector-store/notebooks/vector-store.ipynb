# Vector store playground

Supplied optional demonstration, not graded. Complete the base library first.
No persistence is needed here. Texts are synthetic examples written for this activity.
The initial MiniLM download requires internet; subsequent runs reuse the temporary cache.


```python
from pathlib import Path
from tempfile import gettempdir
from sentence_transformers import SentenceTransformer
from vector_store import Document, VectorStore

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2", device="cpu",
    cache_folder=str(Path(gettempdir()) / "compint" / "10.2.vector-store"),
)
store = VectorStore(model)
```

```python
documents = [
    Document("Python is a programming language used for data analysis.", {"programming", "python"}),
    Document("NumPy provides numerical arrays for scientific computing.", {"programming", "python"}),
    Document("Cats spend much of their day sleeping.", {"animals"}),
    Document("Dogs enjoy walks and playing outside.", {"animals"}),
    Document("A fresh salad can include tomatoes and lettuce.", {"food"}),
    Document("A telescope helps us observe distant stars."),
]
store.add(documents)
```

## Change the query and tags

Use `None` or `set()` for no filter. Multiple tags require all tags.
Scores are cosine similarities, not probabilities.


```python
query = "How can I work with numerical data in Python?"
tags = None
k = 3
for document, score in store.search(query, k=k, metadata=tags):
    print(f"{score:.4f} | {document.text} | {sorted(document.metadata)}")
```

```python
for document, score in store.search(query, k=3, metadata={"programming", "python"}):
    print(f"{score:.4f} | {document.text}")
```
