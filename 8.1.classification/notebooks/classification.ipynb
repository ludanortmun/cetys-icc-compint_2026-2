# 8.1 Classification: Iris

Predict a flower's species using its four numeric measurements. Use an 80/20 split, LogisticRegression, and accuracy. Add new code cells immediately after each marked instruction cell; keep supplied cells intact. See the README for setup and scoring.

Iris has 150 flowers and three classes: `0 = setosa`, `1 = versicolor`, `2 = virginica`. The feature order is sepal length, sepal width, petal length, petal width, all in centimeters. Species codes are categories, not quantities.


```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split

iris = load_iris()
X = iris.data
y = iris.target
print("Feature matrix:", X.shape, "Labels:", y.shape)
print("Features:", iris.feature_names)
print("Species:", iris.target_names)
print("First flower:", X[0], "Species code:", y[0])
```

## 1. Split the dataset — 30 points

**Add a new code cell immediately below this cell.** Call `train_test_split` with the original `X` and `y` together, `test_size=0.20`, `random_state=42`, and other arguments at their defaults. Store the outputs in `X_train`, `X_test`, `y_train`, and `y_test`, in that order.

Keep all four original features and the correct row-label pairs: 120 training examples and 30 test examples. Do not change `X` or `y`.


```python
# Criterion 1: paired 80/20 split (30 points).
_original_X, _original_y = load_iris(return_X_y=True)
_expected_split = train_test_split(
    _original_X, _original_y, test_size=0.20, random_state=42
)
assert X_train.shape == (120, 4)
assert X_test.shape == (30, 4)
assert y_train.shape == (120,)
assert y_test.shape == (30,)
for _actual, _expected in zip(
    (X_train, X_test, y_train, y_test), _expected_split
):
    np.testing.assert_array_equal(_actual, _expected)
print("Split checks passed.")
```

## 2. Train the classifier — 40 points

**Add a new code cell immediately below this cell.** Create `model` as a `LogisticRegression(max_iter=1000)` instance with all other arguments at their defaults. Call `fit` using only `X_train` and `y_train`.

The model learns from labeled training examples. Keep the test examples reserved for evaluation. `max_iter` is the maximum number of training iterations; the fit may finish earlier.


```python
# Criterion 2: LogisticRegression fitted on training data (40 points).
assert isinstance(model, LogisticRegression)
assert model.max_iter == 1000
_reference_model = LogisticRegression(max_iter=1000)
_reference_model.fit(_expected_split[0], _expected_split[2])
assert model.get_params() == _reference_model.get_params()
assert model.n_features_in_ == 4
np.testing.assert_array_equal(model.classes_, [0, 1, 2])
np.testing.assert_allclose(model.coef_, _reference_model.coef_, rtol=1e-6, atol=1e-8)
np.testing.assert_allclose(
    model.intercept_, _reference_model.intercept_, rtol=1e-6, atol=1e-8
)
print("Training checks passed.")
```

## 3. Predict and calculate accuracy — 30 points

**Add a new code cell immediately below this cell.** Call `model.predict(X_test)` and store the returned array as `predictions`. Calculate `accuracy` with `accuracy_score(y_test, predictions)` using default arguments. Display it as a percentage using a print statement.

Store the score as a fraction between zero and one. Accuracy is the number of correct test predictions divided by 30. Do not hardcode a score or predictions; no minimum accuracy is required.


```python
# Criterion 3: test predictions and accuracy (30 points).
assert isinstance(predictions, np.ndarray)
assert predictions.shape == (30,)
np.testing.assert_array_equal(predictions, model.predict(X_test))
assert set(predictions.tolist()).issubset({0, 1, 2})
_correct = sum(int(actual == predicted) for actual, predicted in zip(y_test, predictions))
assert np.isscalar(accuracy)
assert 0.0 <= accuracy <= 1.0
np.testing.assert_allclose(accuracy, _correct / len(y_test), rtol=0, atol=1e-12)
np.testing.assert_allclose(
    accuracy, accuracy_score(y_test, predictions), rtol=0, atol=1e-12
)
print("Prediction and accuracy checks passed.")
```
