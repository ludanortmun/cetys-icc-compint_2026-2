This is where a random seed comes in. Remember that computers cannot produce fully random output, so they use pseudo-random number generators (PRNG) instead. These are pseudo-random because, given the same starting conditions, they will always produce the same result. Normally, by using different starting conditions for each call to the PRNG (for example, the specific timestamp at any given moment), it can *appear* to yield random results.

However, for reproducibility purposes, we may want to opt-out of this randomness temporarily. By choosing a specific random seed, we can ensure that every time we call the PRNG, the result will be the same. Let's start by defining our random seed.

```python
RANDOM_SEED = 42
```
