### Part 3 - Remove non alphabetic and whitespace characters

### Part 4 - Normalizing whitespace

As you can see, because we replaced all non space or alphabetic characters with spaces, the new results can contain duplicate spaces. 

```python
multiple_spaces = [doc for doc in reviews_alpha if '  ' in doc]
print(f"Number of documents with multiple spaces: {len(multiple_spaces)}")

for i in range(10):
    print(multiple_spaces[i])
```

Let's fix this.

**Instructions: On a new cell immediately below this one, create a new variable named `reviews_normalized` with all items from `reviews_alpha` after space normalization. In this case, space normalization refers to collapsing any occurrence of multiple spaces into a single space. For example, `wow    loved this place` must be `wow loved this place`**

```python
for i in range(10):
    display(reviews_normalized[i])
```

```python
assert reviews_normalized is not None, "reviews_normalized is None"
assert len(reviews_normalized) == len(
    reviews_alpha), "Length of reviews_normalized does not match length of reviews_alpha"
for r in reviews_normalized:
    assert not re.match(
        r'\s{2,}', r), "Some documents in reviews_normalized contain multiple consecutive spaces"
```
